```python
import pandas as pd

df = pd.read_csv(
    "../assets/prioritization_candidates.csv"
)

df
```

## Score transparente


```python
df["risk_safety"] = 6 - df["risk"]

dimensions = [
    "business_value",
    "technical_feasibility",
    "data_readiness",
    "risk_safety",
    "time_to_value",
]

df["priority_score"] = (
    df[dimensions].mean(axis=1)
)

df.sort_values(
    "priority_score",
    ascending=False,
)
```
